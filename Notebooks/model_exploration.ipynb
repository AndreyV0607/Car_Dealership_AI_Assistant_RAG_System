### Gemma4 time performance on device
Making use of `time` would be use to see if `gemma4-e2b` runs decently on the hardware

```python
#from ollama import chat 
#import time

start = time.perf_counter()

response = chat(
    model = "gemma4:e2b-it-q4_K_M",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Write a short joke about saving RAM."},
    ],
    options={
        "num_predict": 80,
        "temperature": 0.7,
        "context_window": 4096

    }
)

elapsed = time.perf_counter() - start

print(f"Response", response.message.content)

print(f"Tiempo de ejecución: {elapsed:0.4f} segundos")
```
